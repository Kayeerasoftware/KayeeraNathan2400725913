```python

plt.style.use('seaborn-v0_8-darkgrid')

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Distribution of Sales
sns.histplot(df['Sales'], bins=30, kde=True, ax=axes[0, 0])
axes[0, 0].set_title('Distribution of Sales')
axes[0, 0].set_xlabel('Sales')
axes[0, 0].set_ylabel('Frequency')

# 2. Distribution of Profit
sns.histplot(df['Profit'], bins=30, kde=True, ax=axes[0, 1])
axes[0, 1].set_title('Distribution of Profit')
axes[0, 1].set_xlabel('Profit')
axes[0, 1].set_ylabel('Frequency')

# 3. Sales by Product Category (Bar Chart)
category_sales = df.groupby('ProductCategory')['Sales'].sum().reset_index()

sns.barplot(
    data=category_sales,
    x='ProductCategory',
    y='Sales',
    ax=axes[1, 0]
)

axes[1, 0].set_title('Total Sales by Product Category')
axes[1, 0].set_xlabel('Product Category')
axes[1, 0].tick_params(axis='x', rotation=20)
axes[1, 0].set_ylabel('Total Sales')

# 4. Relationship between Sales and Profit (Scatter Plot)
sns.scatterplot(
    data=df,
    x='Sales',
    y='Profit',
    ax=axes[1, 1],
    alpha=0.7,
    color='teal'
)

axes[1, 1].set_title('Sales vs Profit')
axes[1, 1].set_xlabel('Sales')
axes[1, 1].set_ylabel('Profit')

plt.tight_layout()
plt.show()
```
